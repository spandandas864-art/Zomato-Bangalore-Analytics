### Feature Engineering

Creating has_online_order and has_table_booking features

```python
# Convert 'online_order' to numerical (0 or 1)
df['has_online_order'] = df['online_order'].apply(lambda x: 1 if x == 'Yes' else 0)

# Convert 'isTable_booked' to numerical (0 or 1)
df['has_table_booking'] = df['isTable_booked'].apply(lambda x: 1 if x == 'Yes' else 0)

print("Created 'has_online_order' and 'has_table_booking' features.")
display(df[['online_order', 'has_online_order', 'isTable_booked', 'has_table_booking']].head())
```

Creating num_cuisines feature

```python
# Create a feature for the number of cuisines offered
df['num_cuisines'] = df['cuisines'].apply(lambda x: len(str(x).split(',')))

print("Created 'num_cuisines' feature.")
display(df[['cuisines', 'num_cuisines']].head())
```

Dropping original online_order and isTable_booked columns

```python
df.drop(columns=['online_order', 'isTable_booked'], inplace=True)
print("Dropped original 'online_order' and 'isTable_booked' columns.")
display(df.head())
```
