### Hypothesis testing

```python
from scipy import stats

# Separate ratings for restaurants with and without table booking
ratings_with_table_booking = df[df['has_table_booking'] == 1]['rating']
ratings_without_table_booking = df[df['has_table_booking'] == 0]['rating']

# Perform independent samples t-test
t_stat, p_value = stats.ttest_ind(ratings_with_table_booking, ratings_without_table_booking, equal_var=False) # Assuming unequal variances

print(f"T-statistic: {t_stat:.3f}")
print(f"P-value: {p_value:.3f}")

alpha = 0.05
if p_value < alpha:
    print(f"\nSince the p-value ({p_value:.3f}) is less than the significance level ({alpha}), we reject the null hypothesis.\nThis suggests a statistically significant difference in ratings between restaurants with and without table booking.")
else:
    print(f"\nSince the p-value ({p_value:.3f}) is greater than the significance level ({alpha}), we fail to reject the null hypothesis.\nThis suggests no statistically significant difference in ratings between restaurants with and without table booking.")
```

```python
from scipy.stats import pearsonr

# Calculate Pearson correlation coefficient and p-value
correlation, p_value = pearsonr(df['average_cost_for_two'], df['rating'])

print(f"Pearson Correlation Coefficient: {correlation:.3f}")
print(f"P-value: {p_value:.3f}")

alpha = 0.05
if p_value < alpha:
    print(f"\nSince the p-value ({p_value:.3f}) is less than the significance level ({alpha}), we reject the null hypothesis.\nThis suggests a statistically significant linear relationship between average cost for two and restaurant rating.")
else:
    print(f"\nSince the p-value ({p_value:.3f}) is greater than the significance level ({alpha}), we fail to reject the null hypothesis.\nThis suggests no statistically significant linear relationship between average cost for two and restaurant rating.")
```

```python
from scipy.stats import f_oneway

# Group data by restaurant_location and get the 'average_cost_for_two' for each group
location_costs = [group['average_cost_for_two'].values for name, group in df.groupby('restaurant_location')]

# Perform ANOVA test
f_statistic, p_value = f_oneway(*location_costs)

print(f"F-statistic: {f_statistic:.3f}")
print(f"P-value: {p_value:.3f}")

alpha = 0.05
if p_value < alpha:
    print(f"\nSince the p-value ({p_value:.3f}) is less than the significance level ({alpha}), we reject the null hypothesis.\nThis suggests a statistically significant difference in average cost for two across different restaurant locations.")
else:
    print(f"\nSince the p-value ({p_value:.3f}) is greater than the significance level ({alpha}), we fail to reject the null hypothesis.\nThis suggests no statistically significant difference in average cost for two across different restaurant locations.")
```

```python
from scipy.stats import f_oneway

# Group data by restaurant_location and get the 'rating' for each group
location_ratings = [group['rating'].values for name, group in df.groupby('restaurant_location')]

# Perform ANOVA test
f_statistic, p_value = f_oneway(*location_ratings)

print(f"F-statistic: {f_statistic:.3f}")
print(f"P-value: {p_value:.3f}")

alpha = 0.05
if p_value < alpha:
    print(f"\nSince the p-value ({p_value:.3f}) is less than the significance level ({alpha}), we reject the null hypothesis.\nThis suggests a statistically significant difference in ratings across different restaurant locations.")
else:
    print(f"\nSince the p-value ({p_value:.3f}) is greater than the significance level ({alpha}), we fail to reject the null hypothesis.\nThis suggests no statistically significant difference in ratings across different restaurant locations.")
```

```python
from scipy.stats import f_oneway

# Group data by num_cuisines and get the 'rating' for each group
cuisines_ratings = [group['rating'].values for name, group in df.groupby('num_cuisines')]

# Perform ANOVA test
f_statistic, p_value = f_oneway(*cuisines_ratings)

print(f"F-statistic: {f_statistic:.3f}")
print(f"P-value: {p_value:.3f}")

alpha = 0.05
if p_value < alpha:
    print(f"\nSince the p-value ({p_value:.3f}) is less than the significance level ({alpha}), we reject the null hypothesis.\nThis suggests a statistically significant difference in ratings across different numbers of cuisines.")
else:
    print(f"\nSince the p-value ({p_value:.3f}) is greater than the significance level ({alpha}), we fail to reject the null hypothesis.\nThis suggests no statistically significant difference in ratings across different numbers of cuisines.")
```
